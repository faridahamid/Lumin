```python
dataset_path = "/content/last-v-17-1"

import os
print("Dataset path:", dataset_path)
print("Contents:", os.listdir(dataset_path))
```

```python
import glob

def check_split(split):
    imgs = glob.glob(f"{dataset_path}/{split}/images/*")
    lbls = glob.glob(f"{dataset_path}/{split}/labels/*.txt")

    empty = 0
    seg_like = 0

    for f in lbls:
        txt = open(f).read().strip()
        if txt == "":
            empty += 1
        else:
            for line in txt.splitlines():
                if len(line.split()) > 5:
                    seg_like += 1
                    break

    print(f"\n--- {split.upper()} ---")
    print("Images:", len(imgs))
    print("Labels:", len(lbls))
    print("Empty label files:", empty)
    print("Segmentation-like labels:", seg_like)

check_split("train")
check_split("valid")
check_split("test")
```

```python
import os

def clean_segmentation_labels(split):
    label_files = glob.glob(f"{dataset_path}/{split}/labels/*.txt")
    removed = 0

    for f in label_files:
        lines = open(f).read().strip().splitlines()
        for line in lines:
            if len(line.split()) > 5:
                os.remove(f)
                removed += 1
                break

    print(f"Removed {removed} segmentation-label files from {split}")

clean_segmentation_labels("train")
clean_segmentation_labels("valid")
clean_segmentation_labels("test")
```

```python
def remove_empty_labels(split):
    label_files = glob.glob(f"{dataset_path}/{split}/labels/*.txt")
    removed = 0

    for f in label_files:
        if open(f).read().strip() == "":
            os.remove(f)
            removed += 1

    print(f"Removed {removed} empty-label files from {split}")

remove_empty_labels("train")
remove_empty_labels("valid")
remove_empty_labels("test")
```

```python
def remove_images_without_labels(split):
    img_files = glob.glob(f"{dataset_path}/{split}/images/*")
    removed = 0

    for img in img_files:
        base = os.path.splitext(os.path.basename(img))[0]
        label_path = f"{dataset_path}/{split}/labels/{base}.txt"

        if not os.path.exists(label_path):
            os.remove(img)
            removed += 1

    print(f"Removed {removed} images without labels from {split}")

remove_images_without_labels("train")
remove_images_without_labels("valid")
remove_images_without_labels("test")
```

```python
check_split("train")
check_split("valid")
check_split("test")
```

```python
from collections import Counter, defaultdict

label_files = glob.glob(f"{dataset_path}/train/labels/*.txt")

class_counts = Counter()
image_counts = defaultdict(int)

for file in label_files:
    present = set()
    with open(file) as f:
        for line in f:
            cls = int(line.split()[0])
            class_counts[cls] += 1
            present.add(cls)
    for cls in present:
        image_counts[cls] += 1

print("Annotations per class:")
for k, v in sorted(class_counts.items()):
    print(f"Class {k}: {v}")

print("\nImages per class:")
for k, v in sorted(image_counts.items()):
    print(f"Class {k}: {v} images")
```

```python
from ultralytics import YOLO

# Start from pretrained YOLO weights (not a previously trained run)
model = YOLO("yolo26s.pt")

model.train(
    data="/content/last-v-17-1/data.yaml",

    # Training control
    epochs=60,           # upper bound
    patience=10,         # early stopping
    batch=16,
    imgsz=640,

    # Learning rate (lower than default)
    lr0=5e-4,            # default ~1e-3 â†’ softened
    lrf=1e-5,

    # Freeze backbone from the beginning
    freeze=10,

    # Stability
    optimizer="AdamW",   # better for fine-grained tuning
    weight_decay=0.01,

    # Bookkeeping
    name="from_start_frozen_lowLR",
    save=True
)
```

```python
import os
import shutil

# === CONFIG ===
RUN_NAME = "from_start_frozen_lowLR"
RUN_DIR = f"runs/detect/{RUN_NAME}"
WEIGHTS_DIR = f"{RUN_DIR}/weights"

# Destination on Google Drive
DRIVE_DIR = "/content/drive/MyDrive/Lumin/yolo_runs/from_start_frozen_lowLR"
os.makedirs(DRIVE_DIR, exist_ok=True)

# Files to copy
files_to_copy = {
    "best.pt": "lumin_best.pt",
    "last.pt": "lumin_last.pt",
    "args.yaml": "args.yaml",
    "results.csv": "results.csv",
}

for src_name, dst_name in files_to_copy.items():
    src_path = os.path.join(WEIGHTS_DIR if src_name.endswith(".pt") else RUN_DIR, src_name)
    dst_path = os.path.join(DRIVE_DIR, dst_name)
    shutil.copy(src_path, dst_path)

print("âœ… YOLO outputs saved to Google Drive:")
for f in os.listdir(DRIVE_DIR):
    print(" -", f)
```
